```python
from sklearn.linear_model import LinearRegression
import numpy as np
modelo = LinearRegression()

X = np.array([
[60, 2, 1, 1, 10, 5.0, 450, 3, 1, 0],
[85, 3, 2, 2, 6, 3.0, 600, 7, 1, 1],
[45, 1, 1, 1, 15, 8.0, 300, 2, 0, 0],
[120, 3, 3, 2, 4, 2.0, 850, 10, 1, 1],
[150, 4, 3, 3, 2, 1.5, 950, 12, 1, 1],
[70, 2, 2, 1, 8, 4.0, 500, 5, 1, 0]
])

y = np.array([320, 520, 240, 780, 980, 410])

modelo.fit(X, y)

novo_imovel = np.array([[100, 3, 2, 2, 5, 2.5, 700, 8, 1, 1]])
preco_estimado = modelo.predict(novo_imovel)

print(preco_estimado)
```
